```python
file = '/mnt/hdd_bob/syy/adipose/atac/res/VIB_10xmultiome_2_WS3000F/entropy_filtered_bc_df.tsv'
entropy_df = pd.read_csv(file, sep=',', index_col=0)
```

```python
entropy_df.head(2)
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(entropy_df['Entropy'], entropy_df['mle_lambda'], alpha=0.5, 
           c= entropy_df['P_closed_state'])
ax.set_xlabel('Entropy')
ax.set_ylabel('MLE Lambda')
# add legend for color
cbar = plt.colorbar(ax.collections[0], ax=ax)
cbar.set_label('P_closed_state')
plt.title('Entropy vs MLE Lambda colored by P_closed_state')
```

```python
union_cell_df_file='/mnt/hdd_bob/syy/adipose/atac/res/VIB_10xmultiome_2_WS3000F/_cell_palling_comparison/Union_3set_atac_BCs.tsv'
Union_cell_df = pd.read_csv(union_cell_df_file, sep='\t')
Union_cell_df.index = Union_cell_df['atac_bc'] + '-1'

Union_cell_df.head(2)
```

```python
col_list = [  'total_fragments', 'rna_pass_CR', 'frag_overlap_CR_1stRound_peaks']
a = entropy_df.merge(Union_cell_df[col_list], left_index=True, right_index=True, how='left')
a['FRIP_1stRound'] = a['frag_overlap_CR_1stRound_peaks'] / a['total_fragments']
```

```python
a.head()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(a['Entropy'], a['total_fragments'], alpha=0.5, 
           c= a['P_closed_state'])
ax.set_xlabel('Entropy')
ax.set_ylabel('Total Fragments')
# add legend for color
cbar = plt.colorbar(ax.collections[0], ax=ax)
cbar.set_label('P_closed_state')
plt.title('Entropy vs Total Fragments colored by P_closed_state')
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(entropy_df['P_closed_state'], bins=100, color='skyblue', edgecolor='black')
ax.set_xlabel('P_closed_state')
ax.set_ylabel('Frequency')
plt.title('Distribution of P_closed_state')
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter( np.log(a['total_fragments']), np.log(a['Entropy']), alpha=0.5, 
           c = a['P_closed_state'])
ax.set_xlabel('total_fragments (log scale)')
ax.set_ylabel('Entropy')
# add legend for color
cbar = plt.colorbar(ax.collections[0], ax=ax)
cbar.set_label('P_closed_state')
plt.title('P_closed_state vs Entropy colored by Total Fragments')
```

```python
a['DNAdebris']  = a['P_closed_state'] < 0.5

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter( np.log(a['total_fragments']), a['Entropy'], alpha=0.5, 
           c = a['DNAdebris'].map({True: 'red', False: 'blue'}))
ax.set_xlabel('total_fragments (log scale)')
ax.set_ylabel('Entropy')
ax.legend(handles=[plt.Line2D([0], [0], marker='o', color='w', label='Multiplet', markerfacecolor='red', markersize=10),
                   plt.Line2D([0], [0], marker='o', color='w', label='Singlet', markerfacecolor='blue', markersize=10)], 
          bbox_to_anchor=(1.1, 1), loc='upper left')
plt.title('Multiplet vs Singlet colored by Total Fragments')
```

```python
excessive_reads_bc = a[a['DNAdebris']].sort_values('P_closed_state', ascending=False).index.tolist()
len(excessive_reads_bc)
```
