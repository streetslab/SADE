```python
import numpy as np
import math

from scipy.stats import norm
```

```python
def prob_ZTPD(size, critical_region=0.05, lam=0.1):
    mean_ZTPD = lam / (1 - math.exp(-lam))
    var_ZTPD =  mean_ZTPD * (1 + lam - mean_ZTPD)

    # compute 1- beta, beta is the power funciton of a size
    _beta = 1 - norm.cdf(critical_region, loc= mean_ZTPD, scale=math.sqrt(  var_ZTPD / size  ) )

    return _beta
```

```python
def prob(size, critical_region=0.05, lam=0.1):
    accum = 0
    
    _lambda = size * lam
    _upperlimit = np.floor(critical_region * size)
    
    for i in range(0, int(_upperlimit)+1):
        prob = (_lambda**i) * np.exp(-_lambda) / math.factorial(i)
        accum += prob
        
    return accum

def p0( lam=0.1):
    p0 = np.exp( - lam)
    return p0
```
