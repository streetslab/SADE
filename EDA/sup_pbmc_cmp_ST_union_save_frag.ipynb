### Load fragment file that have reads correspoinding to the Union BCs

```python
union_bc_subdir = os.path.join(entropy_dir, '_cell_calling_comparison')
# union_bc_frag_file = os.path.join(union_bc_subdir, 'othermethod_filtered_fragments.tsv.gz')
# tbx = ps.TabixFile(union_bc_frag_file)
union_bc_sam_file = os.path.join(union_bc_subdir, 'sorted_tagged_union_bc_fragments.bam')


```

```python
com_dir = os.path.join(union_bc_subdir, 'cmp_ST')

common_bc_bam_file = os.path.join(com_dir, 'common_bc_fragments.bam')
entropy_bam_file = os.path.join(com_dir, 'entropy_only_bc_fragments.bam')
tss_bam_file = os.path.join(com_dir, 'tss_only_bc_fragments.bam')
non_bam_file = os.path.join(com_dir, 'non_bc_fragments.bam')
```

```python
for out, id in zip([tss_bam_file, entropy_bam_file, common_bc_bam_file], 
                   ['tss_only', 'entropy_only', 'both']):
    sam = ps.AlignmentFile(union_bc_sam_file, 'rb')
    outfile = ps.AlignmentFile(out, "wb", template=sam)
    for read in sam.fetch(until_eof=True):
        #  if the read has the tag id
        if read.get_tag('ST') == id:
                outfile.write(read)

    outfile.close()
    sam.close()
```

```python
sam = ps.AlignmentFile(union_bc_sam_file, 'rb')
outfile = ps.AlignmentFile(non_bam_file, "wb", template=sam)
for read in sam.fetch(until_eof=True):
    #  if the read has the tag id
    if read.get_tag('ST') == 'none':
            outfile.write(read)

outfile.close()
sam.close()
```
